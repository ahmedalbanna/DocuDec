```python
import json
from typing import Union

def format_value(key: str, value: Union[list, dict, str, int]) -> str:
    """Formats a single key-value pair and returns a string."""
    output = f"{key.capitalize()}:\n"
    if isinstance(value, list):
        for item in value:
            output += f"- {item}\n"
    elif isinstance(value, dict):
        for d_key, d_value in value.items():
            output += format_value(d_key, d_value)  # Recursive call
    else:
        output += f"{value}\n"
    return output


def get_project_data(project_name: str, data: dict) -> dict:
    """Retrieves project data from the input data."""
    return data.get(project_name, {})  # Return empty dict if not found


def format_project_data(project_name: str, data: dict) -> str:
    """Formats project data into a string."""
    project_data = get_project_data(project_name, data)
    if not project_data:
        return "No project data found." #Changed print to return statement to conform with return string

    output = "Project Data:\n"
    for key, value in project_data.items():
        output += format_value(key, value)
    return output


with open("project.json", "r") as f:
    data = json.load(f)
    formatted_data = format_project_data("project", data)
    print(formatted_data)


```
