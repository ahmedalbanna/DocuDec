# Uploads the given file to Gemini.

```python
def upload_to_gemini(path, mime_type=None):
  """Uploads the given file to Gemini.

  See https://ai.google.dev/gemini-api/docs/prompting_with_media
  """
  file = genai.upload_file(path, mime_type=mime_type)
  print(f"Uploaded file '{file.display_name}' as: {file.uri}")
  return file
```

```python
# Project Title
generation_config = {
  "temperature": 1,
  "top_p": 0.95,
  "top_k": 40,
  "max_output_tokens": 500,
  "response_schema": content.Schema(
    type = content.Type.OBJECT,
    enum = [],
    required = ["projectTitle", "description", "elements"],
    properties = {
      "projectTitle": content.Schema(
        type = content.Type.STRING,
      ),
      "description": content.Schema(
        type = content.Type.STRING,
      ),
      "elements": content.Schema(
        type = content.Type.OBJECT,
        enum = [],
        required = ["automation", "focus", "audience"],
        properties = {
          "automation": content.Schema(
            type = content.Type.BOOLEAN,
          ),
          "focus": content.Schema(
            type = content.Type.STRING,
          ),
          "audience": content.Schema(
            type = content.Type.STRING,
          ),
        },
      ),
    },
  ),
  "response_mime_type": "application/json",
}
model = genai.GenerativeModel(
  model_name="gemini-1.5-pro",
  generation_config=generation_config,
  system_instruction="You are a Graduation Project Agent, assisting students in crafting comprehensive and well-structured graduation project reports. Your role includes guiding the student through the various sections of the report, such as project information, content development, structure, and format. You will help create a cohesive narrative that addresses the problem statement, elaborates on the research questions, and summarizes the literature review. Additionally, you will support the student in detailing the project’s methodology, design, implementation, testing, and results analysis. Ensure adherence to academic standards, including tone, style, and formatting requirements.",
)
```

```python
import json

def save_history_json(history, filename="chat_history.json"):
    with open(filename, "w") as f:
        json.dump(history, f, indent=4)  # Save with pretty printing

def load_history_json(filename="chat_history.json"):
    try:
        with open(filename, "r") as f:
            return json.load(f)
    except FileNotFoundError:
        return []  # Return an empty list if the file doesn't exist
```
