```python
import os
from pathlib import Path
import subprocess
import tempfile
from jinja2 import Environment, FileSystemLoader

def generate_and_convert_report(template_path, output_docx_path, data):
    """Generates a report from a Jinja2 template, fills it with data, and converts to docx."""
    try:
        # 1. Render the Jinja2 template with data
        env = Environment(loader=FileSystemLoader(os.path.dirname(template_path)))
        template = env.get_template(os.path.basename(template_path))
        rendered_markdown = template.render(**data)

        # 2. Write rendered Markdown to a temporary file
        with tempfile.NamedTemporaryFile(mode="w", suffix=".md", delete=False) as temp_md_file:
            temp_md_file.write(rendered_markdown)
            temp_md_filepath = temp_md_file.name

        # 3. Use Pandoc to convert the temporary Markdown file to docx
        try:
            subprocess.run(
                [
                    "pandoc",
                    #
                    "--from",
                    "markdown",
                    "--to",
                    "docx",
                    "-o",
                    output_docx_path,
                    temp_md_filepath,
                ],
                check=True,
            )
        except subprocess.CalledProcessError as e:
            print(f"Pandoc execution failed: {e}")
            raise  # Re-raise the exception after logging

        #Clean up the temporary markdown file
        os.remove(temp_md_filepath)
    except Exception as e:
        print(f"An error occurred: {e}")
        raise  # Re-raise the exception for handling by a calling function

# Example usage:

data = {
    "project_title": "Automated Report Generation",
    "student_name": "Ahmed Murtada Albanna",
    "date": "October 26, 2024",
    # ... other data ...
}

template_path = "ReportTemplate.md"  # Path to your Jinja2 template
output_docx_path = "output.docx"  # Path for the output docx file

generate_and_convert_report(template_path, output_docx_path, data)

```
